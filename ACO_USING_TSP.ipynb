```python
import random
n = int(input("Enter number of cities: "))
print("Enter the distance matrix row by row:")
distances = []
for i in range(n):
    row = list(map(int, input().split()))
    distances.append(row)
num_ants = n        # number of ants = number of cities
num_iterations = 50 # how many times algorithm runs
alpha = 1      # pheromone importance
beta = 5       # distance importance
rho = 0.5      # evaporation rate
Q = 100        # pheromone deposit factor
pheromone = [[1 for _ in range(n)] for _ in range(n)]
def probability(i, j, visited):
    tau = pheromone[i][j] ** alpha
    eta = (1 / distances[i][j]) ** beta if distances[i][j] > 0 else 0
    return tau * eta if j not in visited else 0
def construct_solution():
    start = random.randint(0, n-1)
    tour = [start]
    visited = set(tour)
    while len(tour) < n:
        i = tour[-1]
        probs = [probability(i, j, visited) for j in range(n)]
        total = sum(probs)
        if total == 0: break
        # roulette wheel selection
        r = random.uniform(0, total)
        s = 0
        for j, p in enumerate(probs):
            s += p
            if s >= r:
                tour.append(j)
                visited.add(j)
                break
    tour.append(start)  # return to start
    return tour
def tour_length(tour):
    return sum(distances[tour[i]][tour[i+1]] for i in range(len(tour)-1))
best_tour = None
best_length = float("inf")
for _ in range(num_iterations):
    all_tours = [construct_solution() for _ in range(num_ants)]
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)
    for tour in all_tours:
        L = tour_length(tour)
        if L < best_length:
            best_length = L
            best_tour = tour
        for i in range(len(tour)-1):
            a, b = tour[i], tour[i+1]
            pheromone[a][b] += Q / L
print("Yashwanth's 1WN24CS133 Output: ")
print("Best tour found:", best_tour)
print("Tour length:", best_length)
```
