```python
import random

# Take dynamic input from user
n = int(input("Enter number of items: "))
items = []
for i in range(n):
    value = int(input(f"Enter value of item {i+1}: "))
    weight = int(input(f"Enter weight of item {i+1}: "))
    items.append((value, weight))

capacity = int(input("Enter knapsack capacity: "))
population_size = int(input("Enter population size: "))
generations = int(input("Enter number of generations: "))

# Fitness function
def fitness(gene):
    total_value, total_weight = 0, 0
    for bit, (value, weight) in zip(gene, items):
        if bit == 1:
            total_value += value
            total_weight += weight
    return total_value if total_weight <= capacity else 0

# Generate random gene
def random_gene():
    return [random.randint(0,1) for _ in items]

# Mutation: flip a random bit
def mutate(gene):
    i = random.randint(0, len(gene)-1)
    gene[i] = 1 - gene[i]
    return gene

# Main loop
population = [random_gene() for _ in range(population_size)]
for gen in range(generations):
    population = sorted(population, key=fitness, reverse=True)
    best = population[0]
    print(f"Gen {gen+1}: Best gene {best}, Value = {fitness(best)}")

    # Reproduce with mutation
    new_pop = [best[:]]
    while len(new_pop) < population_size:
        child = mutate(best[:])
        new_pop.append(child)
    population = new_pop

print("\nOptimal Solution Found:")
print("Gene:", best)
print("Value:", fitness(best))
```
